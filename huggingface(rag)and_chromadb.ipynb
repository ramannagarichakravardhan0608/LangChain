```python
from langchain_huggingface import HuggingFaceEmbeddings
```

```python
emb = HuggingFaceEmbeddings(model="all-MiniLM-L6-v2")
```

```python
pip install chromadb
```

```python
from langchain_community.vectorstores import Chroma
```

```python
# create a database
```

```python
from langchain_core.documents import Document
```

```python
data = [Document(page_content="langchain is a framework create by the llm based on the application",metadata={"topic":"langchain","difficult":"easy"}),
        Document(page_content="langchain  is contain four compnents langchain,langsmith,langgraph,",metadata={"topic":"langchain","difficult":"medium"}),
        Document(page_content="ml strands for machine learning",metadata={"topic":"ml","difficult":"easy"}),
        Document(page_content="ml is divided into 3 types ",metadata={"topic":"ml","difficult":"easy"}),]
```

```python
db = Chroma.from_documents(documents=data, embedding=emb,persist_directory="db1")
db.persist()
```

```python
#acess all the data in the databases
db._collection.get()
```
