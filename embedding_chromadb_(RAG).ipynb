```python
from langchain_community.document_loaders import PyPDFLoader
```

```python
loader = PyPDFLoader("/content/GenAI - General FAQs.pdf")
doc = loader.load()
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
```

```python
doc_split = RecursiveCharacterTextSplitter(
    chunk_size = 300,
    chunk_overlap = 20
)
```

```python
final_split = doc_split.split_documents(doc)
```

```python
from langchain_community.vectorstores import Chroma
```

```python
database = Chroma(
    collection_name="geniai_FAQs",
    embedding_function=emb_llm,
    persist_directory="chroma_db.db"
)
```

```python
database.add_documents(doc)
```

```python
query = "what is the gen ai"
```

```python
vec = database.as_retriever(search_type="similarity", search_kwargs={"k":2})
```

```python
vec.invoke("what is the AI")
```
