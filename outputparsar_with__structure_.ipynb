```python
from typing import Optional,Literal
```

```python
class Schema(BaseModel):
  name:str = Field(default="chakravardhan",description="exctract my name")
  age:int = Field(description="extract the age if > 20 ",gt=20)#gt-grethan than
  skills:list[str] = Field(description="extract the skills")
  cons:Optional[str] = Field(default = "na",description="extract the cons if existed")
  Position:Literal["Ceo","Hr","triner"]= Field(default="hr")
```

```python
Schema(age =21,name ="chakraavrdhan",skills=['agentic ai','deep learning'],cons="lots of break",Position="Hr")
```

```python
class Schema(BaseModel):
  name:str = Field(description="exctract my name")
  age:Optional[int] = Field(default=0,description="extract the age its contain tha age")
  role:str = Field(description="extract my role")
  summary:str = Field(description="extract my summary")
```

```python
fm = model.with_structured_output(Schema)
```
