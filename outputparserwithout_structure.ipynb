```python
from langchain_core.output_parsers import PydanticOutputParser
```

```python
from pydantic import BaseModel
```

```python
class Student_project_review(BaseModel):
  name:str
  batch:int
  Branch:str
  project_review_status:str
  total_mark: int
  feedback:str
```

```python
parser = pydantic_output_parser = PydanticOutputParser(pydantic_object=Student_project_review)
```

```python
from langchain_core.prompts import ChatPromptTemplate
```

```python
final_prompt = ChatPromptTemplate.from_messages([
    ("system","you have a 5+ experiance in data extraction"),
    ("human",'''Extract the student information from the given text
    text:
    {input}
    {format_instruction}''')
])
```

```python
final_prompt
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI
```

```python
import os
from google.colab import userdata
gem = userdata.get("gemini")
os.environ["GOOGLE_API_KEY"] = gem
```

```python
model = ChatGoogleGenerativeAI(
    model = "gemini-2.5-flash"
)
```

```python
chain = final_prompt | model | parser
```

```python
chain.invoke({
    "input":'''student name is chakravardhan,he is from 410 batch,
                  from Hyd Branch.project review status is yes ans she got total marks 40/100.
                  feedback : she need to improve technical skills and communcation skills.''',
    "format_instruction" : parser.get_format_instructions()
})
```
