```python
## download tokenizer and model
```

```python
# Load model directly
from transformers import AutoTokenizer, AutoModelForCausalLM

tokenizer = AutoTokenizer.from_pretrained("deepseek-ai/deepseek-coder-1.3b-instruct")
model = AutoModelForCausalLM.from_pretrained("deepseek-ai/deepseek-coder-1.3b-instruct")
```

```python
tokenizer.save_pretrained("/content/sample_data/deeepseek")
model.save_pretrained("/content/sample_data/deeepseek")
```

```python
# create the pipeline
```

```python
from transformers import pipeline
```

```python
deep_hf = pipeline(model=model,tokenizer=tokenizer,task="text-generation")
```

```python
from langchain_huggingface import ChatHuggingFace,HuggingFacePipeline
```

```python
deep_cm = ChatHuggingFace(llm=model)
```

```python
response = deep_cm.invoke([("system","give the output in a profisinal way "),("user","tell me about rohitsharma")])
```

```python
response.content
```

```python
model = HuggingFacePipeline(pipeline=deep_hf)
```

```python
import time
```

```python
x = time.time()
respone=model.invoke("tell me about the rohit sharma")
y = time.time()
print(y-x)
```
