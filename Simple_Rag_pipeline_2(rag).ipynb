```python
import os
from google.colab import userdata
ok=userdata.get('open1')
os.environ['OPENAI_API_KEY']=ok
```

```python
# 1 step document loader
from langchain_community.document_loaders import PyPDFLoader

doc=PyPDFLoader("/content/HR-Policies-Manuals.pdf")
document=doc.load()
```

```python
# 2 text-splitter -chunkings
from langchain_text_splitters import RecursiveCharacterTextSplitter

rts=RecursiveCharacterTextSplitter(separators=["\n\n","\n"," ",],chunk_size=500,chunk_overlap=10)

chunks=rts.split_documents(document)
```

```python
pip install chromadb
```

```python
# 3 and 4 -- embdedding ans storing
from langchain_community.vectorstores import Chroma
from langchain_openai import OpenAIEmbeddings

embed=OpenAIEmbeddings(model="text-embedding-3-small")

vdb=Chroma.from_documents(documents=chunks,embedding=embed,persist_directory="db")
vdb.persist()
```

```python
retriver=vdb.as_retriever(search_kwargs={"k":3})
```

```python
retriver.invoke("hi how are you")
```

```python
def f1(data):
  l=[]
  for y in data:
    l.append(y.page_content)

  return "\n\n".join(l)
```

```python
from langchain_core.runnables import RunnableLambda,RunnableParallel,RunnablePassthrough,RunnableSequence
```

```python
r2=RunnableLambda(f1)
```

```python
chain1=RunnableSequence(retriver,r2)
```

```python
chain2=RunnableParallel({"context":chain1,"question":RunnablePassthrough()})
```

```python
chain2.invoke("sdadd")
```

```python
from ast import Str
from langchain_core.prompts import ChatPromptTemplate,HumanMessagePromptTemplate,SystemMessagePromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain_openai import ChatOpenAI

cpt=ChatPromptTemplate.from_messages([SystemMessagePromptTemplate.from_template("you are a helpful hr assistance who is having 4 years of experinece"),
                                 HumanMessagePromptTemplate.from_template("""Answer the question based on the below given context and if there is no context then please return i dont know :
                                 context:{context}
                                 question:{question}""")])

model=ChatOpenAI(model="gpt-5.2-2025-12-11")

sto=StrOutputParser()


```

```python
chain3=RunnableSequence(cpt,model,sto)
```

```python
rag_pipeline=RunnableSequence(chain2,chain3)
```

```python
rag_pipeline.invoke("how many leave i can take in a month")
```

```python
rag_pipeline.invoke("notice period")
```
