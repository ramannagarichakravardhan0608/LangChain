```python
from langchain_text_splitters import RecursiveCharacterTextSplitter,Language,CharacterTextSplitter
```

```python
doc = '''
def fun_1(n,m):
  print(n+m)
  print(n-m)
fun_1(6,7)
def fun_2(n,m):
  print(n*m)
  print(n/m)
fun_2(5,6)
fun_2(7,8) '''
```

```python
code_split = RecursiveCharacterTextSplitter.from_language(
    language = Language.PYTHON,
    chunk_size=30,
    chunk_overlap=0,

)
```

```python
code_split.split_text(doc)
```

```python
code_split = CharacterTextSplitter(
    separator=" ",
    chunk_size=30,
    chunk_overlap=0,

)
```

```python
code_split.split_text(doc)
```

```python
from transformers import AutoTokenizer
```

```python
tokens = AutoTokenizer.from_pretrained("deepseek-ai/DeepSeek-R1")
```

```python
token_split = RecursiveCharacterTextSplitter.from_huggingface_tokenizer(
    tokenizer=tokens,
    chunk_size=200,
    chunk_overlap=30
)
```

```python
token_split.split_text(doc)
```

```python
len(token_split.split_text(doc))
```

```python
token_split.split_text(doc)[0]
```
